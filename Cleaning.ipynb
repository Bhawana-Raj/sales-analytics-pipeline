```python
import pandas as pd

df = pd.read_excel('sales_raw_500.xlsx')
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
# Removes Duplicates 
df = df.drop_duplicates()
```

```python
df
```

```python
# Fix date columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors = 'coerce')
```

```python
# Drop NA or blank values from Order_date column

df = df.dropna(subset=['Order_Date'])
```

```python
df
```

```python
# Add new Column

df['Profit'] = df['Sales']-df['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
df
```

```python
df.to_csv('sales_cleaned_data.csv' ,index = False)
```

```python
# Engine build connection between python and sql

import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('sales_cleaned_data.csv')
```

```python
df
```

```python
# Data build connection inside engine so,build ecreate_engine

engine = create_engine('postgresql+psycopg2://postgres:123456@localhost:5432/Sales_db')
print(engine)
```

```python
df.to_sql(name='sales_data', con=engine, if_exists='replace', index=False)
```
